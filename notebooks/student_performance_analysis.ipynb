```python
q1, q3 = df["absences"].quantile([0.25, 0.75])
iqr = q3 - q1
upper = q3 + 1.5 * iqr
print("Upper outlier threshold:", upper)
print("Students above threshold:", (df["absences"] > upper).sum())

plt.figure(figsize=(6, 3))
sns.boxplot(x=df["absences"])
plt.title("Absences")
plt.show()
```

```python
clean = df.copy()

# Remove exact duplicate rows (expected: none)
clean = clean.drop_duplicates()

# Strip stray whitespace in text columns
for col in clean.select_dtypes(include="object").columns:
    clean[col] = clean[col].str.strip()

print(clean.shape)
```

```python
clean = clean.drop(columns=["romantic"])

rename_map = {
    "Medu": "mother_edu", "Fedu": "father_edu",
    "Mjob": "mother_job", "Fjob": "father_job",
    "studytime": "study_time", "traveltime": "travel_time",
    "famrel": "family_relationship", "goout": "go_out",
    "Dalc": "alcohol_weekday", "Walc": "alcohol_weekend",
    "schoolsup": "school_support", "famsup": "family_support",
    "paid": "paid_classes", "higher": "wants_higher_ed",
    "G1": "g1", "G2": "g2", "G3": "g3",
}
clean = clean.rename(columns=rename_map)

print(clean.shape)
clean.columns.tolist()
```

```python
assert clean.isnull().sum().sum() == 0, "Missing values found"
assert clean.duplicated().sum() == 0, "Duplicates found"
assert clean["g3"].between(0, 20).all(), "g3 out of range"
assert clean["age"].between(15, 22).all(), "age out of range"
print("All checks passed")
```

```python
clean.to_csv("../data/processed/student_por_clean.csv", index=False)

check = pd.read_csv("../data/processed/student_por_clean.csv")
print(check.shape)
```
